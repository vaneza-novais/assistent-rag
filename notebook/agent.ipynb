### Como o chat IA funciona normalmente

```python
from langchain.chat_models import init_chat_model

# inicilizar o LLM
llm = init_chat_model(
    model = 'gemini-3.5-flash-lite',
    model_provider = 'google_genai'
)
```

```python
from typing import TypedDict, Annotated, Sequence # classe para criar dicionários
from langgraph.graph import add_messages
from langchain_core.messages import BaseMessage

# 1. definir o state
class AgentState(TypedDict): # armazenar o historico de msg 
    nodes_path: Annotated[
        Sequence[BaseMessage], # Define que o valor guardado na chave deve ser uma sequência (uma lista ou tupla) contendo objetos de mensagem do LangChain.
        add_messages # concatena as novas mensagens ao histórico existente
    ]

# 2. Definir o node (funcões que executam as ações)
def call_llm(state: AgentState):
    llm_result = llm.invoke(
        state['nodes_path'] # passar todo historico de msg
    )
    return {'nodes_path': [llm_result]}
```

```python
from langgraph.graph import StateGraph, START, END

# 3. Criar o stategraph
builder = StateGraph( #classe principal usada pra criar o grafo
    AgentState, #o que transita entre os nodes
    context_schema=None, #estrutura de dependencias externas injetadas
    # garantir que a entrada, o processamento interno e a saída usam exatamente o mesmo formato de dicionário.
    input_schema= AgentState, 
    output_schema= AgentState
)


# 4. Adicionar os nodes ao grafico
builder.add_node('call_llm', call_llm)
builder.add_edge(START, 'call_llm')
builder.add_edge('call_llm', END)

# 5. Compilar o grafo
graph = builder.compile()
```

```python
from langchain.messages import HumanMessage

if __name__ == "__main__":
    # 6. Iniciar o fluxo do grafo
    human_message = HumanMessage('Oi, tudo bem? Prazer, meu nome é Joao. Fale sobre regras de tamanho maximo em cada linha')
    result = graph.invoke({'nodes_path': [human_message]})
    print(result)
```
